```python
import pandas as pd



```

```python
# Load the dataset
# We refer to the file by its exact name: "dining.csv"
df = pd.read_csv("dining.csv")
```

```python
# read in your data in this cell

# Display the first 5 rows to understand the structure
print(df.head())
# Print total rows and columns as (rows, columns)
print("Shape of dataset:", df.shape)

# View summary of column names, non-null counts, and data types
df.info()

```

```python
# print the first 5 rows of your data in this cell

# Keep only the rows for Brooks dining hall
filtered_df = df[df["dining_hall"] == "Brooks"].copy()

# Display the first five Brooks rows
print(filtered_df.head())
```

```python
# Calculate coffee consumption per swipe for each day
filtered_df["average_coffee"] = filtered_df["coffee_consumed"] / filtered_df["swipes"]

# Display the first five rows with the new column
print(filtered_df.head())
```

```python
# Sort from highest to lowest coffee consumption per swipe
filtered_df = filtered_df.sort_values(by="average_coffee", ascending=False)

# Display the first five rows after sorting
print(filtered_df.head())
```

```python
# Save the sorted Brooks data without an extra index column
filtered_df.to_csv("modified_dining.csv", index=False)

print("Saved modified_dining.csv")
```
